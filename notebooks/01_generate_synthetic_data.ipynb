# Marketing Analytics Dataset

## Row grain

Each row represents one customer's interaction with one product through one marketing campaign on one date.

## Dataset objective

The dataset will support:

- marketing KPI analysis
- campaign performance comparison
- customer segmentation
- conversion prediction
- revenue and return-on-ad-spend analysis

## Dataset schema

| Column | Type | Meaning |
|---|---|---|
| interaction_date | date | Date of the interaction |
| customer_id | string | Unique customer identifier |
| product_id | string | Unique product identifier |
| campaign_id | string | Unique marketing campaign identifier |
| customer_age | integer | Customer's age |
| customer_gender | string | Customer's gender category |
| customer_region | string | Customer's region |
| product_category | string | Product category |
| marketing_channel | string | Campaign channel, such as Email or Social Media |
| device_type | string | Device used by the customer |
| unit_price | double | Original price of one product unit |
| discount_percent | double | Discount percentage offered |
| impressions | integer | Number of times the advertisement was displayed |
| clicks | integer | Number of advertisement clicks |
| website_visits | integer | Number of resulting website visits |
| orders | integer | Number of completed orders |
| units_sold | integer | Total number of units purchased |
| ad_spend | double | Advertising cost allocated to the interaction |
| revenue | double | Revenue after applying the discount |

**Import tools and set basic configuration**

```python
import random
from datetime import date, timedelta

# Make the generated data repeatable
SEED = 42
random.seed(SEED)

# Number of rows to generate
NUM_ROWS = 10_000

# Dataset date range
START_DATE = date(2025, 1, 1)
END_DATE = date(2025, 12, 31)

print("Configuration ready")
print("Number of rows:", NUM_ROWS)
print("Date range:", START_DATE, "to", END_DATE)
```

**Define the possible values**

```python
# Customer-related options
GENDERS = ["Male", "Female"]

REGIONS = [
    "Punjab",
    "Sindh",
    "Khyber Pakhtunkhwa",
    "Balochistan",
    "Islamabad"
]

DEVICE_TYPES = [
    "Mobile",
    "Desktop",
    "Tablet"
]

# Product catalogue
PRODUCTS = [
    {"product_id": "P001", "product_category": "Electronics", "unit_price": 50000.0},
    {"product_id": "P002", "product_category": "Clothing", "unit_price": 5000.0},
    {"product_id": "P003", "product_category": "Footwear", "unit_price": 8000.0},
    {"product_id": "P004", "product_category": "Beauty", "unit_price": 3000.0},
    {"product_id": "P005", "product_category": "Accessories", "unit_price": 2500.0}
]

# Marketing campaigns
CAMPAIGNS = [
    {
        "campaign_id": "CMP001",
        "marketing_channel": "Email",
        "click_probability": 0.10,
        "visit_probability": 0.95,
        "order_probability": 0.15,
        "minimum_cpm": 8000,
        "maximum_cpm": 12000
    },
    {
        "campaign_id": "CMP002",
        "marketing_channel": "Social Media",
        "click_probability": 0.14,
        "visit_probability": 0.88,
        "order_probability": 0.10,
        "minimum_cpm": 10000,
        "maximum_cpm": 16000
    },
    {
        "campaign_id": "CMP003",
        "marketing_channel": "Search",
        "click_probability": 0.18,
        "visit_probability": 0.95,
        "order_probability": 0.18,
        "minimum_cpm": 15000,
        "maximum_cpm": 22000
    },
    {
        "campaign_id": "CMP004",
        "marketing_channel": "Display",
        "click_probability": 0.07,
        "visit_probability": 0.82,
        "order_probability": 0.07,
        "minimum_cpm": 5000,
        "maximum_cpm": 9000
    }
]

print("Regions:", len(REGIONS))
print("Products:", len(PRODUCTS))
print("Campaigns:", len(CAMPAIGNS))
```

**Create the customer profiles**

```python
# Reset the random generator for reproducibility
random.seed(SEED)

NUM_CUSTOMERS = 1_000
customers = []

for number in range(1, NUM_CUSTOMERS + 1):
    customer = {
        "customer_id": f"C{number:04d}",
        "customer_age": random.randint(18, 65),
        "customer_gender": random.choice(GENDERS),
        "customer_region": random.choice(REGIONS)
    }

    customers.append(customer)

print("Customers created:", len(customers))

for customer in customers[:5]:
    print(customer)
```

**Create a function for random dates**

```python
def generate_random_date(start_date, end_date):
    total_days = (end_date - start_date).days
    random_number_of_days = random.randint(0, total_days)

    return start_date + timedelta(days=random_number_of_days)
```

```python
for _ in range(5):
    print(generate_random_date(START_DATE, END_DATE))
```

**Generate the interaction records**

```python
# Reset randomness so the data can be reproduced
random.seed(SEED)

interactions = []
used_interaction_keys = set()

# Continue until 10,000 unique interactions are created
while len(interactions) < NUM_ROWS:
    customer = random.choice(customers)
    product = random.choice(PRODUCTS)
    campaign = random.choice(CAMPAIGNS)
    interaction_date = generate_random_date(START_DATE, END_DATE)

    # Date + customer + product + campaign must be unique
    interaction_key = (
        interaction_date,
        customer["customer_id"],
        product["product_id"],
        campaign["campaign_id"]
    )

    # Reject the interaction if this combination already exists
    if interaction_key in used_interaction_keys:
        continue

    # Remember the new unique combination
    used_interaction_keys.add(interaction_key)

    # The customer can see the same advertisement several times
    impressions = random.randint(1, 10)

    # Use the selected campaign's click probability
    if random.random() < campaign["click_probability"]:
        clicks = 1
    else:
        clicks = 0

    # Use the selected campaign's visit probability
    if (
        clicks == 1
        and random.random() < campaign["visit_probability"]
    ):
        website_visits = 1
    else:
        website_visits = 0

    # Use the selected campaign's order probability
    if (
        website_visits == 1
        and random.random() < campaign["order_probability"]
    ):
        orders = 1
    else:
        orders = 0

    # One order can contain between one and three units
    if orders == 1:
        units_sold = random.randint(1, 3)
    else:
        units_sold = 0

    # Select a possible discount
    discount_percent = random.choice([0, 5, 10, 15, 20, 25, 30])

    # Calculate revenue after discount
    revenue = (
        units_sold
        * product["unit_price"]
        * (1 - discount_percent / 100)
    )

    # Use the selected campaign's advertising cost range
    cost_per_thousand_impressions = random.uniform(
        campaign["minimum_cpm"],
        campaign["maximum_cpm"]
    )

    ad_spend = (
        impressions
        / 1000
        * cost_per_thousand_impressions
    )

    # Combine all values into one interaction record
    interaction = {
        "interaction_date": interaction_date,
        "customer_id": customer["customer_id"],
        "product_id": product["product_id"],
        "campaign_id": campaign["campaign_id"],
        "customer_age": customer["customer_age"],
        "customer_gender": customer["customer_gender"],
        "customer_region": customer["customer_region"],
        "product_category": product["product_category"],
        "marketing_channel": campaign["marketing_channel"],
        "device_type": random.choice(DEVICE_TYPES),
        "unit_price": product["unit_price"],
        "discount_percent": float(discount_percent),
        "impressions": impressions,
        "clicks": clicks,
        "website_visits": website_visits,
        "orders": orders,
        "units_sold": units_sold,
        "ad_spend": round(ad_spend, 2),
        "revenue": round(revenue, 2)
    }

    interactions.append(interaction)

print("Interactions created:", len(interactions))
print("Unique interaction keys:", len(used_interaction_keys))

for interaction in interactions[:3]:
    print(interaction)
```

**Convert the data into a Spark DataFrame**

```python
df = spark.createDataFrame(interactions)

print("Number of rows:", df.count())

df.printSchema()

display(df)
```

**Check the funnel **rules****

```python
from pyspark.sql.functions import col

invalid_funnel_rows = df.filter(
    (col("clicks") > col("impressions")) |
    (col("website_visits") > col("clicks")) |
    (col("orders") > col("website_visits")) |
    (col("units_sold") < col("orders"))
)

invalid_count = invalid_funnel_rows.count()

print("Total rows:", df.count())
print("Invalid funnel rows:", invalid_count)

if invalid_count == 0:
    print("Funnel validation passed")
else:
    print("Funnel validation failed")
    display(invalid_funnel_rows)
```

**Check for missing values**

```python
from pyspark.sql.functions import col, sum as spark_sum

missing_values = df.select([
    spark_sum(
        col(column_name).isNull().cast("int")
    ).alias(column_name)
    for column_name in df.columns
])

display(missing_values)
```

**Check for duplicate interactions**

```python
grain_columns = [
    "interaction_date",
    "customer_id",
    "product_id",
    "campaign_id"
]

duplicate_groups = (
    df.groupBy(*grain_columns)
      .count()
      .filter(col("count") > 1)
)

duplicate_count = duplicate_groups.count()

print("Duplicate groups:", duplicate_count)

if duplicate_count == 0:
    print("Duplicate validation passed")
else:
    print("Duplicate validation failed")
    display(duplicate_groups)
```

**Validate the revenue calculation**

Revenue = Units sold × Unit price × Amount remaining after discount

```python
from pyspark.sql import functions as F

df_revenue_check = df.withColumn(
    "expected_revenue",
    F.round(
        F.col("units_sold")
        * F.col("unit_price")
        * (1 - F.col("discount_percent") / 100),
        2
    )
)

invalid_revenue_rows = df_revenue_check.filter(
    F.abs(
        F.col("revenue") - F.col("expected_revenue")
    ) > 0.01
)

invalid_revenue_count = invalid_revenue_rows.count()

print("Invalid revenue rows:", invalid_revenue_count)

if invalid_revenue_count == 0:
    print("Revenue validation passed")
else:
    print("Revenue validation failed")
    display(invalid_revenue_rows)
```

**Save the dataset as a Delta table**

```python
table_name = "marketing_interactions_raw"

(
    df.write
      .format("delta")
      .mode("overwrite")
      .saveAsTable(table_name)
)

print("Table saved successfully:", table_name)
```

**Read the table using SQL**

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT customer_id) AS unique_customers,
    MIN(interaction_date) AS first_date,
    MAX(interaction_date) AS last_date,
    SUM(impressions) AS total_impressions,
    SUM(clicks) AS total_clicks,
    SUM(orders) AS total_orders,
    ROUND(SUM(revenue), 2) AS total_revenue
FROM marketing_interactions_raw;
```

**Calculate the main marketing KPIs**

```python
%sql

SELECT
    ROUND(
        100.0 * SUM(clicks) / NULLIF(SUM(impressions), 0),
        2
    ) AS click_through_rate_percent,

    ROUND(
        100.0 * SUM(website_visits) / NULLIF(SUM(clicks), 0),
        2
    ) AS click_to_visit_rate_percent,

    ROUND(
        100.0 * SUM(orders) / NULLIF(SUM(website_visits), 0),
        2
    ) AS conversion_rate_percent,

    ROUND(
        SUM(revenue) / NULLIF(SUM(ad_spend), 0),
        2
    ) AS return_on_ad_spend,

    ROUND(
        SUM(revenue) / NULLIF(SUM(orders), 0),
        2
    ) AS average_order_value,

    ROUND(
        SUM(ad_spend) / NULLIF(SUM(orders), 0),
        2
    ) AS cost_per_order

FROM marketing_interactions_raw;
```

```python
%sql

SELECT
    campaign_id,
    marketing_channel,

    SUM(impressions) AS total_impressions,
    SUM(clicks) AS total_clicks,
    SUM(website_visits) AS total_visits,
    SUM(orders) AS total_orders,

    ROUND(SUM(revenue), 2) AS total_revenue,
    ROUND(SUM(ad_spend), 2) AS total_ad_spend,

    ROUND(
        100.0 * SUM(clicks) / NULLIF(SUM(impressions), 0),
        2
    ) AS click_through_rate_percent,

    ROUND(
        100.0 * SUM(orders) / NULLIF(SUM(website_visits), 0),
        2
    ) AS conversion_rate_percent,

    ROUND(
        SUM(revenue) / NULLIF(SUM(ad_spend), 0),
        2
    ) AS return_on_ad_spend

FROM marketing_interactions_raw

GROUP BY
    campaign_id,
    marketing_channel

ORDER BY
    return_on_ad_spend DESC;
```

# KPI Definitions

## Dataset grain

Each row represents one customer's interaction with one product through one marketing campaign on one date.

## Click-through rate

**Formula:**

Clicks / Impressions × 100

**Meaning:**  
The percentage of advertisement displays that produced a click.

## Click-to-visit rate

**Formula:**

Website visits / Clicks × 100

**Meaning:**  
The percentage of clicks that successfully became website visits.

## Conversion rate

**Formula:**

Orders / Website visits × 100

**Meaning:**  
The percentage of website visits that produced an order.

The denominator is website visits, not impressions or clicks.

## Return on ad spend

**Formula:**

Revenue / Advertising spend

**Meaning:**  
The revenue generated for every PKR 1 spent on advertising.

ROAS measures revenue, not profit.

## Average order value

**Formula:**

Revenue / Orders

**Meaning:**  
The average revenue generated by one purchase transaction.

One order may contain multiple units.

## Cost per order

**Formula:**

Advertising spend / Orders

**Meaning:**  
The average advertising cost required to generate one order.

Orders ÷ Website visits
